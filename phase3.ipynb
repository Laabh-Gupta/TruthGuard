```python
import importlib
import retrieval.rag_pipeline
importlib.reload(retrieval.rag_pipeline)

from retrieval.rag_pipeline import RAGPipeline

rag = RAGPipeline()
result = rag.query("What machine learning projects has the candidate worked on?")

report = result["hallucination_report"]

# --- Grounding Score ---
print("=" * 60)
print(f"🎯 TRUTHGUARD ANALYSIS")
print("=" * 60)
print(f"Overall Verdict : {report['overall_verdict']}")
print(f"Grounding Score : {report['grounding_percent']}")
print(f"Total Sentences : {report['total_sentences']}")
print(f"  ✅ Supported  : {report['supported_count']}")
print(f"  ⚠️  Weak       : {report['weak_count']}")
print(f"  ❌ Unsupported : {report['unsupported_count']}")

# --- Per sentence breakdown ---
print("\n" + "=" * 60)
print("🔍 SENTENCE BREAKDOWN")
print("=" * 60)
for i, s in enumerate(report["sentences"]):
    icon = "✅" if s["verdict"] == "SUPPORTED" else "⚠️" if s["verdict"] == "WEAK" else "❌"
    print(f"\n[{i+1}] {icon} {s['verdict']} (score: {s['best_score']})")
    print(f"     Sentence : {s['sentence'][:100]}...")
    print(f"     Evidence : {s['best_chunk_text'][:100]}...")

# --- Flag hallucinations ---
if report["unsupported_sentences"]:
    print("\n" + "=" * 60)
    print("🚨 POTENTIAL HALLUCINATIONS DETECTED")
    print("=" * 60)
    for s in report["unsupported_sentences"]:
        print(f"\n  ❌ {s['sentence'][:120]}")
        print(f"     Best similarity was only: {s['best_score']}")
else:
    print("\n✅ No hallucinations detected — answer fully grounded")
```

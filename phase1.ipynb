```python
import sys
sys.path.append("D:/Projects/TruthGuard")  # adjust to your path

from ingestion.pdf_loader import load_pdf, chunk_pages
from retrieval.vector_store import VectorStore
from retrieval.rag_pipeline import RAGPipeline

# --- Step 1: Ingest a PDF ---
pages = load_pdf("Laabh_Gupta.pdf")  # put any PDF in your truthguard folder
chunks = chunk_pages(pages)

# --- Step 2: Store in ChromaDB ---
store = VectorStore()
store.add_chunks(chunks)

# --- Step 3: Ask a question ---
rag = RAGPipeline()
result = rag.query("What is this document about?")

print("\n📄 ANSWER:")
print(result["answer"])

print("\n📚 SOURCES USED:")
for s in result["sources"]:
    print(f"  - {s['source']} | Page {s['page']} | Similarity: {s['similarity']}")
```
